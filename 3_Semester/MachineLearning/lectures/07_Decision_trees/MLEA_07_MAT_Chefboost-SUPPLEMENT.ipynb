As mentioned in the main notebook, this small example here is to illustrate how ID3 and C4.5 can be used with a third party library called `chefboost`. 

Please note, that there have been several bugs in this library in the past and it is not very well maintained. However, we will create a decision tree using categorical data with this library...

Before you can import the library, you have to install it using `pip install chefboost` (it is not maintained by Anaconda, that's why we have to use pip). In general, you can find more information about this library at it's [Github page](https://github.com/serengil/chefboost). 

So, import first:

```python
import pandas as pd
from chefboost import Chefboost as chef
```

Now, let's import another toy dataset, the "churn" data from the lecture. It serves again as a (too) tiny example containing categorical data only.
Please note that the "train" data is the same from the lecture, the "test" data is just one additional record we want to classify (it has no class label, computing an accuracy for that would be meaningless anyway...)

```python
df = pd.read_csv("data/churn_train.csv", sep=";", index_col=0)
df = df.astype(object) # this is necessary since data type 'str' is not handled correctly by chefboost, whyever..
df
```

```python
df = df.astype(object)

config = {'algorithm': 'ID3'} # the original one, if you want, try out C4.5 as well
model = chef.fit(df, config = config, target_label = 'churn')


```

As you can see, the syntax of this is very different from what you are used to when using `scikit-learn`. Additionally, ignore the "Evaluate train set" message, it evaluates on the train set (which we know gives as 100% accuracy

The library stores the decision tree as rules in Python code in the directory `outputs/rules`, so let's check them out (you can open the file with your browser or we can try to import and print them here:

```python
with open('outputs/rules/rules.py', 'r') as f:
    print(f.read())
```

If you carefully compare this code with the plot on page 14 of the lecture slides, you will find out it is the same. Although here last 4 lines seem odd.

Furthermore, we could import the code and call the function:

```python
import outputs.rules.rules as r
r.findDecision(["many", "M", "single", "eco"]) # this is case #1, so we should get a "yes"
```

Now let's test the model with our test data

```python
test = pd.read_csv("data/churn_test.csv", sep=";", index_col=0) # just one record, no ground truth either..
test = test.astype(object)
test
```

```python
prediction = chef.predict(model, param = test.values.tolist()[0]) # also here the syntax is very strange -- you can pass one instance only
                                                                  # if you want to test multiple instances, namely a test set, you need to program a
                                                                  # loop :(
print(prediction)
```
